# IAAIS Environment Check

This notebook reports whether the course environment dependencies and spaCy model are available. It is diagnostic rather than a substitute for the automated tests.

```python
import importlib.util
import sys

required = (
    "jupyter", "ipykernel", "sklearn", "numpy", "pandas",
    "matplotlib", "seaborn", "networkx", "spacy", "nltk",
    "transformers", "torch", "openai", "streamlit", "dotenv",
)
availability = {name: importlib.util.find_spec(name) is not None for name in required}
print("Python:", sys.version.split()[0])
for name, available in availability.items():
    print(f"{name}: {'available' if available else 'missing'}")
```

```python
try:
    import spacy
    nlp = spacy.load("en_core_web_sm")
    print("spaCy model: en_core_web_sm available")
    print(nlp("IAAIS environment is ready.").text)
except Exception as error:
    print("spaCy model: unavailable")
    print(type(error).__name__ + ":", error)
    print("Install with: python -m spacy download en_core_web_sm")
```
